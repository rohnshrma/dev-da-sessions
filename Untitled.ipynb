```python
import random 
```

```python

attempts = 0
max_attempts = 10
secret_number = random.randint(1,101)
won = False
while attempts < max_attempts:
    attempts += 1
    guess = int(input(f"Attempts - {attempts} / {max_attempts}\nEnter your guess : "))
    if  guess  < secret_number:
        print("Too Low! Try High.")
    elif guess > secret_number:
        print("Too High! Try Low.")
    else:
        print(f"Congratulations! You've guessed the correct number : {secret_number} in {attempts} attempts.")
        won = True
        break

if not won == True:
    print(f"Failed to guess the number! The secret number was {secret_number}")
```
