```python
follow_up_count = 0
student_interested = True
max_attempts = 5

while student_interested and follow_up_count < max_attempts:
    response = input("Did the student response? (yes/no/not interested) : ").lower()

    if response == "yes":
        print("Admission Marked. Followup Ends")
        student_interested = False
    elif response == "not interested":
        print("Student Marked as Not Interested. Followup Ends")
        student_interested = False
    elif response == "no":
        follow_up_count += 1
        print(f"No Response. Follow up attempt {follow_up_count} logged")
    else:
        print("Invalid Input")


    if follow_up_count == max_attempts:
        print("Maximum followup attempts reached. stop folowwing up!")
```
