```python
count = 0

n = 122345
while n > 0:
    print(n)
    count += 1
    n = n // 10 # remove last digit


print(count)
```

```python
rev = 0

n = 341342345
while n > 0:
    digit = n % 10
    print(f"Digit : {digit}")
    rev = rev * 10 + digit
    print(f"Rev : {rev}")
    n = n // 10 # remove last digit
    print(f"Number : {n}")
print(f"Reversed :{rev}")
```
